exercício - 15 ⏬

```python
valor_hora = int(input('quanto voce ganha por hora? '))
horas_mes = int(input('quantas horas voce trabalhou esse mês? '))

salario_b = valor_hora * horas_mes
print(f'salário bruto é: ${salario_b}')

ir = (salario_b * 11 / 100)

inss = (ir * 8 / 100)
print(f'quanto voce pagou ao inss: {inss}')

sindicato = (inss * 5 / 100)
print(f'quanto voce pagou ao sindicato: {sindicato}')

salario_l = (salario_b - ir - inss - sindicato)
print(f'seu salario liquido é: {salario_l}')
```
